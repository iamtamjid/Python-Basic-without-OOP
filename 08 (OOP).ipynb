```python
#Write your code here

# TODO
class Shape:
    def __init__(self,name,h,w):
        self.name = name
        self.height = h
        self.width = w

    def area(self):
        if self.name == "Triangle":
            area = 0.5*self.height*self.width
            print("Area:",area)
        elif self.name == "Square":
            area = self.height**2
            print("Area:",area)
        elif self.name == "Rhombus":
            area = (self.height*self.width)/2
            print("Area:",area)
        elif self.name == "Rectangle":
            area = self.height*self.width
            print("Area:",area)
        else:
            area = "Shape unknown"
            print("Area:",area)


# Do not change the following code:
triangle_1 = Shape("Triangle",10,25)
triangle_1.area()
print("==========================")
square_1 = Shape("Square",10,10)
square_1.area()
print("==========================")
rhombus_1 = Shape("Rhombus",18,25)
rhombus_1.area()
print("==========================")
rectangle_1 = Shape("Rectangle",15,30)
rectangle_1.area()
print("==========================")
trapezium_1 = Shape("Trapezium",15,30)
trapezium_1.area()
```

```python
#Write your code here

# TODO
class Calculator:
    def __init__(self,num1,num2,operator):
        self.num1 = num1
        self.num2 = num2
        self.operator = operator
        self.result = None

    def calculate_button(self):
        if self.operator == "+":
            self.result = self.num1 + self.num2
        elif self.operator == "-":
            self.result = self.num1 - self.num2
        elif self.operator == "*":
            self.result = self.num1 * self.num2
        elif self.operator == "/":
            self.result = self.num1 / self.num2
        return self.result

    def details(self):
        print(self.num1,self.operator,self.num2,"=",self.result)

#Do not change the following code:
add = Calculator(10, 20, '+')
val = add.calculate_button()
print("Returned value:", val)
add.details()
print("==================")

sub = Calculator(val, 10, '-')
val = sub.calculate_button()
print("Returned value:", val)
sub.details()
print("==================")

div = Calculator(val, 5, '*')
val = div.calculate_button()
print("Returned value:", val)
div.details()
print("==================")

mult = Calculator(val, 16, '/')
val = mult.calculate_button()
print("Returned value:", val)
mult.details()
print("==================")
```

•	Create a <strong>seating_capacity</strong> method that has default value of 50.<br>
•	The default <strong>fare charge</strong> of any vehicle is seating capacity * 100 and add an extra 10% on full fare as a maintenance charge. So total fare for a vehicle instance will become the final amount = total fare + 10% of the total fare.<br>

<br>
COPY THIS PORTION:<br>
--------------------------------------<br>
<em>#Write your code here<br><br>
#Do not change the following code:</em><br>
School_bus = Vehicle("School Volvo", 180, 12)<br>

--------------------------------------<br>

<strong>OUTPUT:</strong><br>

Vehicle Name: School Volvo<br>
Speed: 180 <br>
Mileage: 12<br>
The seating capacity of a bus is 50 passengers<br>
Total Bus fare is: 5500.0<br>


```python
#Write your code here

# TODO
class Vehicle:
    def __init__(self,n,s,m):
        self.name = n
        self.max_speed = s
        self.mileage = m
    def seating_capacity(self):
        seat_cap = 50
        return seat_cap
    def fare(self):
        total_fare = self.seating_capacity()*100
        final_amount = total_fare+(total_fare*10)/100
        return final_amount
    def details(self):
        print("Vehicle Name:",self.name)
        print("Speed:",self.max_speed)
        print("Mileage:",self.mileage)

#Do not change the following code:
School_bus = Vehicle("School Volvo", 180, 12)
School_bus.details()
print("The seating capacity of a bus is",School_bus.seating_capacity(),"passengers")
print("Total Bus fare is:", School_bus.fare())
```

```python
#Write your code here

# TODO
class Student:
    def __init__(self,name,id,department,grades):
        self.name = name
        self.id = id
        self.department = department
        self.grades = grades
        self.CGPA = None
        self.credit = 3

    def calculate_CGPA(self):
        sub = 0
        for i in range(len(self.grades)):
            sub += self.grades[i]*self.credit

        self.CGPA = sub/(len(self.grades)*3)

    def print_details(self):
        print("--------- Department:",self.department,"---------")
        print("Name:",self.name,",","ID:",self.id)
        print("CGPA:",self.CGPA)
        if 3.80<=self.CGPA<=4.00:
            print("Your academic standing is","'Highest Distinction'")
        elif 3.65<=self.CGPA<=3.79:
            print("Your academic standing is","'High Distinction'")
        elif 3.50<=self.CGPA<=3.64:
            print("Your academic standing is","'Distinction'")
        elif 2.00<=self.CGPA<=3.49:
            print("Your academic standing is","'Satisfactory'")
        elif 0.00<=self.CGPA<=1.99:
            print("You Cannot graduate")
        else:
            print("Wrong Input")

#Do not change the following code:
s1 = Student('Dora', '15995599','CSE', [4,3.7,3.7,4])
s1.calculate_CGPA()
s1.print_details()

s2 = Student('Pingu', '12312322', 'EEE', [1.7,1.3,1.3,1.3,1])
s2.calculate_CGPA()
s2.print_details()

s3 = Student('Bob', '13311331', 'CSE', [2,3,3,3.7,2.7,2.7])
s3.calculate_CGPA()
s3.print_details()
```
